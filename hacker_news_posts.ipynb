```python
df['created_at'] = pd.to_datetime(df['created_at'])
df['created_at'].head()
```

```python
# Now you can extract year, month, day, and hour
df["year"] = df["created_at"].dt.year
df["month"] = df["created_at"].dt.month
df["day"] = df["created_at"].dt.day
df["hour"] = df["created_at"].dt.hour
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
# Separate the 'Ask HN' posts
ask_df = df[df['title'].str.startswith('Ask HN')]        #This line filters the data to find all posts where the title starts with the text 'Ask HN'.
                                                         #df['title'] refers to the column in the dataset that contains the titles of the posts.
                                                          #str.startswith('Ask HN') is a method that checks if the title of the post begins with the string ask_df.

# Separate the 'Show HN' posts
show_df = df[df['title'].str.startswith('Show HN')]         #Similarly, this line filters the dataset to find all posts where the title starts with 'Show HN'.

# Display the first few rows of each
ask_df.head(), show_df.head()

```
